```python

fast_sample_query = """
SELECT tableid, colid, rowid
FROM public.main_tokenized
LIMIT 50000;
"""
df_raw_samples = pd.read_sql(fast_sample_query, conn)

# 2. Count row occurrences in Python (Takes milliseconds!)
table_counts = df_raw_samples['tableid'].value_counts()

# 3. Filter for medium-sized tables (between 150 and 800 token cells)
medium_tables = table_counts[(table_counts >= 150) & (table_counts <= 800)]

# 4. Grab the top 5 unique IDs
sample_ids = medium_tables.index[:5].tolist()
print(f"🚀 Found 5 medium tables instantly in Python: {sample_ids}")
```

```python
import pandas as pd

# 1. Verification Loop for your 5 chosen table IDs
for i, tid in enumerate(sample_ids, start=1):
    print(f"\n" + "="*50)
    print(f"📊 PROCESSING SAMPLE TABLE {i}/5 (ID: {tid})")
    print("="*50)
    
    # Fetch the precise matrix points for this specific table ID
    query = f"""
    SELECT rowid, colid, tokenized
    FROM public.main_tokenized
    WHERE tableid = {int(tid)}
    ORDER BY rowid, colid;
    """
    df_raw = pd.read_sql(query, conn)
    
    if df_raw.empty:
        print(f"⚠️ Table ID {tid} returned no rows. Skipping...")
        continue
        
    # Calculate and display the raw database token counts
    unique_rows = df_raw['rowid'].nunique()
    unique_cols = df_raw['colid'].nunique()
    print(f"📏 Raw Matrix Footprint: {unique_rows} rows x {unique_cols} columns ({len(df_raw)} total tokens)")
    
    try:
        # 2. Pivot the matrix points into an initial 2D layout
        # (Handling duplicate cell records gracefully with pivot_table using max aggregate)
        df_pivot = df_raw.pivot_table(
            index='rowid', 
            columns='colid', 
            values='tokenized', 
            aggfunc='max'
        )
        
        # 3. Transpose to align rows and columns structurally
        df_corrected = df_pivot.T
        
        # 4. Apply your Geometric Bottom-Up Header Finder
        header_idx = None
        for idx in reversed(df_corrected.index):
            row_data = df_corrected.loc[idx]
            valid_cells = [
                str(val).strip() for val in row_data.dropna()
                if str(val).strip() not in ("", "NaN", "nan", "null", "None")
            ]
            if len(valid_cells) > 0:
                header_idx = idx
                break
                
        if header_idx is not None:
            # Assign structural text strings found at the boundary edge as labels
            df_corrected.columns = df_corrected.loc[header_idx].values
            df_corrected = df_corrected.drop(index=header_idx, errors='ignore')
        else:
            print("⚠️ Could not dynamically establish a geometric text header. Using defaults.")
            df_corrected.columns = [f"col_{c}" for c in df_corrected.columns]
            
        # 5. Clean up tracking offsets, index parameters, and empty columns
        df_corrected = df_corrected.drop(index=0, errors='ignore') # Drop rowid 0 anchor if present
        df_corrected = df_corrected.loc[:, df_corrected.columns.notna()]
        df_corrected = df_corrected.dropna(how='all', axis=1)
        df_corrected.index.name = "runner_row_idx"
        
        print(f"✨ Reconstructed Table Dimensions: {df_corrected.shape[0]} rows x {df_corrected.shape[1]} columns")
        print("\n👀 Preview of reconstructed layout:")
        display(df_corrected.head(5))
        
    except Exception as e:
        print(f"❌ Failed to structurally reconstruct Table ID {tid} due to error: {str(e)}")
```

```python
import pandas as pd

for i, tid in enumerate(sample_ids, start=1):
    print("\n" + "="*60)
    print(f"📊 RECONSTRUCTING TABLE {i}/5 (ID: {tid})")
    print("="*60)
    
    query = f"""
    SELECT rowid, colid, tokenized
    FROM public.main_tokenized
    WHERE tableid = {int(tid)}
    ORDER BY rowid, colid;
    """
    df_raw = pd.read_sql(query, conn)
    
    if df_raw.empty:
        continue
        
    # 1. Base Pivot (Keep the natural vertical orientation)
    df_pivot = df_raw.pivot_table(index='rowid', columns='colid', values='tokenized', aggfunc='max')
    
    # Define common header indicators to locate metadata rows
    header_keywords = {'name', 'gender', 'bio', 'other', 'date', 'amount', 'type', 'id', 'title', 'revenue'}
    
    header_idx = None
    
    # 2. Scan every row to check if it matches our header profiles
    for idx, row in df_pivot.iterrows():
        row_words = [str(val).lower().strip() for val in row.dropna()]
        
        # If any word in this row hits our target keyword profile, we've found our header!
        if any(any(kw in word for kw in header_keywords) for word in row_words):
            header_idx = idx
            break
            
    # 3. Apply the headers based on our discovery
    df_final = df_pivot.copy()
    if header_idx is not None:
        print(f"🎯 Header row discovered at rowid: {header_idx}")
        # Assign discovered text values as column names
        df_final.columns = [str(c).strip() for c in df_pivot.loc[header_idx].values]
        # Drop the header row from the data body
        df_final = df_final.drop(index=header_idx, errors='ignore')
    else:
        print("💡 No explicit text header row found. Auto-generating generic column names.")
        df_final.columns = [f"col_{c}" for c in df_pivot.columns]
        
    # 4. Universal Data Polishing
    # Remove any rows where all elements are NaN
    df_final = df_final.dropna(how='all', axis=0)
    # Ensure column headers themselves aren't NaN values
    df_final = df_final.loc[:, df_final.columns.notna()]
    
    print(f"✨ Final Verified Layout: {df_final.shape[0]} rows x {df_final.shape[1]} columns")
    display(df_final.head(5))
```
